```python
import epics
import time
from datetime import datetime

pv_name = "BL531acA5427:cam1:Acquire"
timeout = 20.0
check_interval = 10.0
stuck_time = 0.0

print(f"Starting Python watchdog for {pv_name}...")

while True:
    # Use as_string=True to ensure we can consistently check for 'Acquire'
    status = epics.caget(pv_name, as_string=True)
    
    if status in ['Acquire', '1']:
        stuck_time += check_interval
        if stuck_time >= timeout:
            current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
            print(f"[{current_time}] Camera stuck for {timeout}s. Unsticking...")
            epics.caput(pv_name, 1)
            stuck_time = 0.0
            time.sleep(1) # Pause briefly after the caput
    else:
        stuck_time = 0.0
        
    time.sleep(check_interval)
```
