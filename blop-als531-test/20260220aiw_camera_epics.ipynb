```python
import epics
import numpy as np
import matplotlib.pyplot as plt
```

```python
prefix = "BL531a2A2448:"

# Read data

width = epics.caget(f"{prefix}cam1:ArraySizeX_RBV")
height = epics.caget(f"{prefix}cam1:ArraySizeY_RBV")

image_data = epics.caget(f"{prefix}image1:ArrayData")

# Method 1: Reshape as (height, width, 3) - most common
try:
    image_rgb = np.array(image_data, dtype=np.uint8).reshape(height, width, 3)
    
    plt.figure(figsize=(10, 8))
    plt.imshow(image_rgb)
    plt.title(f'RGB Image ({width}x{height}) - Method 1')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
    
except ValueError:
    # Method 2: If RGB planes are separate (R plane, G plane, B plane)
    image_rgb = np.array(image_data, dtype=np.uint8).reshape(3, height, width)
    # Transpose to (height, width, 3)
    image_rgb = np.transpose(image_rgb, (1, 2, 0))
    
    plt.figure(figsize=(10, 8))
    plt.imshow(image_rgb)
    plt.title(f'RGB Image ({width}x{height}) - Method 2')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```
